# Problema #1: Calculadora de propina inteligente

## Problema

programa que de forma automatica calcule la propina de una cuenta segun la calidad del servicio y dividia el pago final entre varias personas.

## Entrada

- Monto total de cuenta
- Calidad del servicio
- Número de personas

## Proceso

- Determinar el porcentaje segun la calidad del servicio
- Calcular el valor de propina y sumarlo
dividir el total entre la cantidad de personas
- Dividir el total de la cuenta entre la cantidad de personas

## Salida esperada

- El total de la propina a dejar
- El total general a pagar(cuenta + propina)
- Cantidad exacta que debe pagar cada persona

## Pruebas manuales




```python
cuenta = float(input( "Ingrese el total de la cuenta ($): "))
calidad= int(input("Calidad del servicio: 1= Malo (10%), 2= Regular (15%), 3=Exelente(20%)"))
personas = int(input("Ingrese el numero de personas: "))
if calidad == 1:
    propina = cuenta * 0.10
elif calidad == 2:
    propina = cuenta * 0.15
else:
    propina = cuenta * 0.20
total = cuenta + propina
por_persona = total / personas
print(f"El total de la propina es: ${propina:.2f}")
print(f"El total a pagar es: ${total:.2f}")
print(f"Cada persona debe pagar: ${por_persona:.2f}")
```

# Problema #2: Detector de número par/impar

## Problema

crear programa que solicite al usuario un numero entero y determinar de forma automatica si el numero es par o inpar

## Entrada
- nombre del usuario
-  un numero entero

## Proceso

- guardar el numero ingresado por el usuario
- dividir el numero entre 2


## Salida esperada

- mensaje claro que le diga al usuario si es par o inpar

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | | | |
| Caso límite | | | |
| Caso inválido | | | |


```python
#Entrada
nombre_usuario = input("Ingrese su nombre: ") .upper() .strip()
#Proceso
try:
    numero = int(input("Ingrese un número entero: "))
    #Proceso
    if numero % 2 == 0:
        print(f"{nombre_usuario}, el número {numero} es par.")
    else:
        print(f"{nombre_usuario}, el número {numero} es impar.")
except ValueError:
    print("Error: Ingrese un número entero válido.")
    exit()


```

# Problema #3: Clasificador de edad

## Problema

Crear programa que solicite al usuario su edad y determine de forma automatica a que categoria pertenece niño, adolescente, adulto, adulto mayor.
## Entrada

- Nombre del usuario
- Edad de la persona

## Proceso

- Recibir la edad y evaluar la categoria a la que pertenece
- Si es menor de 12 años: Niño
- Si esta entre 12 y 17 años: Adolescente
- Si esta entre 18 y 64 años: Adulto
- Si tiene mas de 65 años: Adulto mayor

## Salida esperada

- mensaje claro en pantalla indicando a cada persona a la categoria que pertenece

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | | | |
| Caso límite | | | |
| Caso inválido | | | |


```python
#Entrada
nombre_usuario = input("Ingrese su nombre: ") .upper() .strip()
edad = int(input("Ingrese su edad: "))
#Proceso
if edad < 12:
  categoria = "niño (a)"
elif edad >= 12 and edad <= 17:
  categoria = "adolescente"
elif edad >= 18 and edad <= 64:
  categoria = "adulto"
else:
  categoria = "adulto mayor"
#Salid
print(f"{nombre_usuario}, usted es un {categoria}.")
```
